```python
from langchain_community.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Load PDF file (provide your own path)
pdf_path = "resume_ng.pdf"  # Replace with your PDF path

# Load PDF and split into pages
loader = PyPDFLoader(pdf_path)
pages = loader.load_and_split()
print(f"Total pages loaded: {len(pages)}")

# Define text splitter with chunk size and overlap
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)

# Split pages into chunks
chunks = text_splitter.split_documents(pages)
print(f"Total chunks created: {len(chunks)}")

# Show sample chunks
print("=== Sample chunk 1 ===")
print(chunks[0].page_content[:500])  # Print first 500 chars

print("\n=== Sample chunk 2 ===")
print(chunks[1].page_content[:500])
```
